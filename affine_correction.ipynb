```python
# Affine correction, check every file in the atlas folder and make sure they have the same affine
atlas_folder = r"C:\github\dog_brain_toolkit\Atlas\Dog\Nitzsche"
import os
import nibabel as nib
import numpy as np

# main affine
main_affine_file = r"C:\github\dog_brain_toolkit\Atlas\Dog\Nitzsche\brain2mm.nii.gz"
main_img = nib.load(main_affine_file)
main_affine = main_img.affine

# list all *2mm.nii.gz files in the atlas folder
for filename in os.listdir(atlas_folder):
    if filename.endswith('2mm.nii.gz'):
        filepath = os.path.join(atlas_folder, filename)
        img = nib.load(filepath)
        data = img.get_fdata()
        current_affine = img.affine
        # copy the main affine if different
        if not np.array_equal(current_affine, main_affine):
            print(f"Correcting affine for {filename}")
            new_img = nib.Nifti1Image(data, main_affine)
            nib.save(new_img, filepath)
        
```
